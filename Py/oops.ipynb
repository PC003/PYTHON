```python
from abc import abstractmethod
from math import pi
class Shape:
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self,side):
       self.side=side

    def area(self):
        return self.side*self.side

class Circle(Shape):
    def __init__(self,radius):
       self.side=radius

    def area(self):
        return pi*self.side*self.side


circle=Circle(4)
sq=Square(4)
circle.area()
sq.area()
```

```python
class FileManger:
    def __init__(self,file):
        self.file=file
    
    def read_file(self):
        with open(self.file,'r')as f:
            content=f.read()
        return content
    
f=FileManger('dest.txt')
print(f.read_file())
```

```python
#Custom iterator
class CountDown:
    def __init__(self,start):
        self.start=start
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.start<0:
            raise StopIteration
        self.start -=1
        return self.start+1
count=CountDown(3)
for num in count:
    print(num)
       
```

```python
def countdown_generator(n):
    start=1
    while start < n:
        if start%2==0:
            yield start
        start +=1

for num in countdown_generator(5):
    print(num)
```

```python
def numbers(n):
    for i in range(1, n + 1):
        yield i

def squares(nums):
    for num in nums:
        yield num ** 2

def filter_even(nums):
    for num in nums:
        if num % 2 == 0:
            yield num

# Chain generators
result = filter_even(squares(numbers(5)))
for num in result:
    print(num)
```

```python
def max_subset(list):
    dp=[0,0,0,0,0,0]
   
    dp[1]=list[0]
    for i in range(2,6):
        if list[i-1]+dp[i-1]>dp[i-1]:
            dp[i]=list[i]+dp[i-1]
    return dp[4]

max_subset([2,-3,5,-8,7])
```

```python
def dp_practice(a,b):
    n=len(a)
    dp=[0]*n
    dp[0]=2
    dp[1]=8
    for i in range(2,n):
       dp[i]=max(max(a[i],b[i])+dp[i-2],dp[i-1])
    
    return dp[n-1]

dp_practice([2,3,4,-8,2],[-5,8,3,1,-4])
```
