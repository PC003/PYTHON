```python
df=pd.read_excel('flight_price.xlsx')
```

```python
df.head()
```

```python
df.info()
```

```python
#let clean the date_0f_journey ,by seperatingit  as day,month,year
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_hrs']=df['Arrival_Time'].str.split(':').str[0]
```

```python
df.head()
```

```python
df['Arrival_min']=df['Arrival_Time'].str.split(' ').str[0].str[3:5]
```

```python
df.head()
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df = df.rename(columns={'hrs': 'Arrival_hrs','min':'Arrival_min'})
df.head()
```

```python
df['Dept_hrs']=df['Dep_Time'].str.split(':').str[0]
df['Dept_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df[['Date','Month','Year','Arrival_hrs','Arrival_min','Dept_min','Dept_hrs']]=df[['Date','Month','Year','Arrival_hrs','Arrival_min','Dept_min','Dept_hrs']].astype(int)
```

```python
df['Duration_hrs']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```

```python
df['Duration_min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0]
```

```python
df.head()
```

```python
df['Duration_min']=df['Duration_min'].fillna('0')
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,np.nan:1})
```

```python
df['Total_Stops']=df['Total_Stops'].fillna(1)
```

```python
df['Total_Stops']=df['Total_Stops'].astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
#for remaining catogorical varible we can use nominal encoding 
from sklearn.preprocessing import OneHotEncoder

encoder=OneHotEncoder()

encoded=encoder.fit_transform(df[['Airline']]).toarray()

encoded_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
encoded_df=pd.concat([df['Airline'],encoded_df],axis=1)
encoded_df.head()
```
