MISSING VALUES
1.Missing Completely at random-No reason for missing value
2.Missing at Random- Missing values are systematically related to the observed data
3.Missing data not at random-dependant on multiple factores

```python
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
#simple way is just delete missing value row -but it will loose alot of data
df.dropna().shape
```

```python
#Column wise would be better 
df.dropna(axis=1).head()
```

Imputation Missing values

```python
##Using mean to fill missing value 
df['age_mean']=df['age'].fillna(df['age'].mean())
```

Mean imputation works well when we have normally distributed data

```python
df[['age_mean','age']]
```

2.Median value imputation - when we have outliers like not normal distrubution

```python
df['age_median']=df['age'].fillna(df['age'].median())
```

```python
df[['age_mean','age_median','age']]
```

3.Median imputation-when we have catagorical varible

```python
mode_value=df[df['embarked'].notna()]['embarked'].mode()[0]
```

```python
mode_value
```

```python
df['embarked_mode']=df['embarked'].fillna(mode_value)
```

```python
df[['embarked','embarked_mode']]
```
