## PorterStemmer


```python
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
```

```python
words=[
    "elephant",
    "computer",
    "chocolate",
    "beautiful",
    "sunflower",
    "pineapple",
    "notebook",
    "football",
    "birthday",
    "building",
    "hospital",
    "airplane",
    "backpack",
    "umbrella",
    "mountains"
]
```

```python
for word in words:
    print(word+" --> "+stemming.stem(word))
```

## Regex Stemmer class

NLTK hase this class with the help of which we can easily implement Regular regex expression stemmer algorithms it basically takdes a sinle regular expression and removes any prefix or suffix that matches the expression let us see an example.

```python
from nltk.stem import RegexpStemmer
reg=RegexpStemmer('ing$|s$|e$|able$', min=4)
## $ing ->for start ,just ing ->removes every ing , ing$ removes in last 
w1=[
    "running",
    "jumping",
    "playing",
    "reading",
    "writing",
    "smiling",
    "talking",
    "dancing",
    "singing",
    "walking",
    "making",
    "taking",
    "coming",
    "driving",
    "hopeful",
    "readable",
    "washable",
    "drinkable",
    "portable",
    "comfortable"
]
for word in w1:
    print(reg.stem(word))
```

## Snowblow stemmer

```python
from nltk.stem import SnowballStemmer
snow=SnowballStemmer('english')

```

```python
for word in w1:
    print(word+" -->"+snow.stem(word))
```

#### Work for some word but not for every word in dictionary so next techinques used is lemmatization 

##### like stemming the output we will get after lemmatization is lemma which is a root word rather than root stem the output of stemming
#### after lemmatization we will be getting a valid word that means the same thing ,used in Q&A ,chatbots


```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wnl=WordNetLemmatizer()
```

```python
for word in w1:
    print(word+" --> "+wnl.lemmatize(word,pos='n'))
```

```python
for word in w1:
    print(word+" --> "+wnl.lemmatize(word,pos='v'))
```
