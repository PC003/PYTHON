```python
import pandas as pd
import numpy as np
```

```python
movie_data=pd.read_csv('./train.csv')
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer

ss=SnowballStemmer('english')
```

```python
corpus=[]

import re

for i in range(0,len(movie_data)):
    review=re.sub('[^a-zA-z]',' ',movie_data['synopsis'][i])
    review=review.lower()
    review=review.split()
    review=[ss.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer=TfidfVectorizer(max_features=15000,ngram_range=(1,2))


```

```python
X=corpus
y=movie_data['genre']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.3,random_state=42)
```

```python
X_train=vectorizer.fit_transform(X_train)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

y_train=le.fit_transform(y_train)
```

```python
from sklearn.linear_model import LogisticRegression

lr=LogisticRegression(max_iter=3000,
    class_weight='balanced')

model1=lr.fit(X_train,y_train)
```

```python
x_test=vectorizer.transform(X_test)
y_test=le.transform(y_test)
```

```python
y_pred=model1.predict(x_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.svm import LinearSVC
lsv=LinearSVC()
model2=lsv.fit(X_train,y_train)
y_pred2=model2.predict(x_test)
```
