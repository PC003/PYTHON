```python
def reading(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            print(f"File has this: \n{content}")
    except FileNotFoundError:
        print(f"File not found: {filename}")

reading(filename='example.txt')
```

```python
def write_file(lines,filename):
    try:
        with open(filename,'a')as file:
            for line in lines:
                file.write(line)
                file.write('\n')
    except FileNotFoundError:
        print(f"File not found:{filename}")
lines=['Also i like f1','i am piyush ','i am studdent ']
write_file(lines,'dest.txt')
reading('dest.txt')
```

```python
def copy_file(source,dest):
    try:
        content=''
        with open(source,'r')as file:
            content+=file.read()
        with open(dest,'w')as file:
            file.write(content)
    except FileNotFoundError:
        print(f"file not found{source}")
copy_file('dest.txt','newcopy.txt')
```

```python
import datetime

def log_message(message, filename='activity.log'):
    timestamp = datetime.datetime.now().isoformat()
    with open(filename, 'a') as file:
        file.write(f'[{timestamp}] {message}\n')

# Test
log_message('This is a log message.')
```

```python
#finding and replacing the text
def replac(prev,next):
    with open('dest.txt','r')as file:
        content=file.read()
        new_content=content.replace(prev,next)
        with open('dest.txt','w')as file:
            file.write(new_content)
replac('i','we')
```

```python
#reading file in reverse

def read_reverse(file):
    with open(file,'r')as file1:
      content=file1.readlines()
      for line in reversed(content):
         print(line)
read_reverse('example.txt')
```

```python
def merged_file(files):
    with open('merged.txt','w')as m:
        for file in files:
            with open (file,'r')as f:
                content=f.read()
                m.write(content)

files=['dest.txt','example.txt']
merged_file(files)
```

```python
import csv
def read_csv(file):
    with open(file,'r')as f:
      reader=csv.DictReader(f)
      data=list(reader)
      return data
read_csv('data.csv')        
```

```python
import json
def read_json(file):
    with open(file,'r')as f:
        content=f.read()
        reader=json.loads(content)
        data=list(reader)
        return data
read_json('data.json')
```
