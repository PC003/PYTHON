```python
import pandas as pd
import matplotlib.pyplot as ptl
import numpy as np
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

```

```python
X,y=load_iris(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,classification_report
gauss=GaussianNB()
gauss.fit(X_train,y_train)
y_pred=gauss.predict(X_test)
print('Accuracy',accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df.head(4)
```

```python
df['time']=np.where(df['time'].str.contains('Dinner'),1,0)
```

```python
tips_X=df.drop(columns=['sex','smoker','day','size','time'])
tips_y=df['time']
```

```python
tips_X_train,tips_X_test,tips_y_train,tips_y_test=train_test_split(tips_X,tips_y,test_size=0.25,random_state=42)
gauss1=GaussianNB()
gauss1.fit(tips_X_train,tips_y_train)
y_pred1=gauss1.predict(tips_X_test)
print('Accuracy',accuracy_score(tips_y_test,y_pred1))
print(classification_report(tips_y_test,y_pred1,zero_division=1.0))
```
