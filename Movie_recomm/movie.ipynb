```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('IMDB-Movie-Dataset(2023-1951).csv')
df.info()
```

```python
df.head(5)
```

```python
#drop unnamed col it not useful
df.drop(df.columns[df.columns.str.contains('unnamed', case=False)], axis=1, inplace=True)
```

```python
df.head(4)
```

```python
df['movie_id']=df['movie_id'].str.replace('tt','')
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()

genre_encoded = pd.DataFrame(
    mlb.fit_transform(df['genre']),
    columns=mlb.classes_,
    index=df.index
)

df = pd.concat([df, genre_encoded], axis=1)
```

```python
df.drop('genre',axis=1,inplace=True)
```

```python
#now genre will issur as it contains mulitple genre
df['genre'] = df['genre'].apply(lambda x: [i.strip() for i in x.split(',')])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus = ["This is the first document.", "This document is the second document."]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```
