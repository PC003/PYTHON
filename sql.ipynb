```python
import sqlite3

# Connect to database
conn = sqlite3.connect('test1.db')
curr = conn.cursor()

# Create table
def create(conn, curr):
    curr.execute('''
        Create Table IF Not Exists employee(
            id Integer Primary Key,
            name Text Not Null,
            age Integer,
            department Text
        )
    ''')
    conn.commit()

create(conn, curr)

# Insert data
def insert(emp, curr, conn):
    curr.executemany('''
        Insert into employee(name, age, department) Values(?, ?, ?)
    ''', emp)
    conn.commit()

emp = [
    ("Piyush", 30, "Computer"),
    ("Rohit", 21, "Combat"),
    ("Nihar", 22, "Sport"),
]
insert(emp, curr, conn)

# Query data
def query(curr, conn):
    for row in curr.execute("Select name, age, department From employee Order by age"):
        print(row)

query(curr, conn)

# Close connection
conn.close()
```
