```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
```

```python
iris
```

```python
X=pd.DataFrame(iris['data'],columns=['sepal length in cm','sepal width (cm)','petal length (cm)','petal width (cm)'])
y=iris['target']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

## In decision we dont need any standardization 

```python
from sklearn.tree import DecisionTreeClassifier

treeClassi=DecisionTreeClassifier()
treeClassi.fit(X_train,Y_train)

```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeClassi,filled=True)
```

## This can lead to overfitting so we do prunning Post or Pre
# Post Prunning ,set max_depth =value etc good for small dataset but pre is like hyperparameter tunning so it will good for big datasets


```python
from sklearn.metrics import classification_report
```

```python

y_pred=treeClassi.predict(X_test)
print(classification_report(Y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

treeClassi1=DecisionTreeClassifier(max_depth=2)
treeClassi1.fit(X_train,Y_train)
y_pred1=treeClassi1.predict(X_test)
print(classification_report(Y_test,y_pred1))
```

### Pre Prunning

```python
params={
    'criterion':['gini', 'entropy', 'log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV

grid=GridSearchCV(treeClassi,param_grid=params,cv=5,scoring='accuracy')
grid.fit(X_train,Y_train)

```

```python
grid.best_params_
```

```python
grid.best_score_
```

### Diabetes Dataset


```python
from sklearn.datasets import load_diabetes
df=load_diabetes()
print(df['DESCR'])
```

```python
print(df)
```

```python
data=pd.DataFrame(df['data'],columns=['age','sex','bmi','bp','s1','s2','s3','s4','s5','s6'])
```

```python
cr=data.copy()
cr['target']=pd.DataFrame(df['target'])
cr.head(4)
```

```python
cr.corr()
```

```python
X_reg=data.drop(columns=['s1','s2'],axis=1)
y_reg=df['target']
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10,16))
sns.heatmap(data=cr.corr(),annot=True,cmap='coolwarm',fmt='.1f')
##show that target and sex have very less correlation
```

```python
from sklearn.model_selection import train_test_split
X_train_reg,X_test_reg,Y_train_reg,Y_test_reg=train_test_split(X_reg,y_reg,test_size=0.30,random_state=42)
```

```python

from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor(max_depth=3,random_state=42,criterion='squared_error',splitter='best',max_features='sqrt')
model.fit(X_train_reg,Y_train_reg)
y_pred_reg=model.predict(X_test_reg)
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score
print(mean_absolute_error(Y_test_reg,y_pred_reg))
print(root_mean_squared_error(Y_test_reg,y_pred_reg))
print(r2_score(Y_test_reg,y_pred_reg))
```

```python
params1={
    'criterion':['squared_error', 'friedman_mse', 'absolute_error','poisson'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['sqrt','log2']
}
from sklearn.model_selection import GridSearchCV

grid1=GridSearchCV(DecisionTreeRegressor(),param_grid=params1,cv=5,scoring='r2')
grid1.fit(X_train,Y_train)
```
