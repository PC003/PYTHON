```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_redundant=0,n_clusters_per_class=1)

```

```python
pd.DataFrame(X)
```

```python
y
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0],y=pd.DataFrame(X)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.svm import SVC
svc=SVC(kernel='linear')
svc.fit(X_train,y_train)

```

```python
y_pred=svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
X_new,y_new=make_classification(n_samples=1000,n_features=2,n_classes=2,n_redundant=0,n_clusters_per_class=2)
```

```python
sns.scatterplot(x=pd.DataFrame(X_new)[0],y=pd.DataFrame(X_new)[1],hue=y)
```

```python
X_train_new,X_test_new,y_train_new,y_test_new=train_test_split(X_new,y_new,test_size=0.25,random_state=42)
```

```python
l=SVC(kernel='linear')
l.fit(X_train_new,y_train_new)
y_pred0=l.predict(X_test_new)
print(classification_report(y_test_new,y_pred0))
```

```python
rbf=SVC(kernel='rbf')
rbf.fit(X_train_new,y_train_new)
y_pred1=rbf.predict(X_test_new)
print(classification_report(y_test_new,y_pred1))
```

## Hyperparameter tunning


```python
from sklearn.model_selection import GridSearchCV
```

```python
params_grid={
     'C':[0.1,1,10,100,1000],
     'gamma':[1,0.1,0.01,0.001,0.0001],
     'kernel':['rbf']

}
grid=GridSearchCV(SVC(),param_grid=params_grid,refit=True,cv=5,verbose=3)
grid.fit(X_train_new,y_train_new)
```

```python
grid.best_params_
```

```python
#lets get tips dataset
df=sns.load_dataset('tips')
```

```python
df['smoker']=np.where(df['smoker'].str.contains('Yes'),1,0)
```

```python
df['time']=np.where(df['time'].str.contains('Dinner'),1,0)
```

```python
df['sex']=np.where(df['sex'].str.contains('Male'),1,0)
```

```python
df.tail(4)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[4])],remainder='passthrough')
```

```python
df=ct.fit_transform(df)
```

```python
import sys
np.set_printoptions(threshold=sys.maxsize)
df=pd.DataFrame(df)
df
```

```python
X=df.drop(3,axis=1)
y=df[3]
y
```

```python
X_train1,X_test1,y_train1,y_test1=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
#SVR
from sklearn.svm import SVR
svr=SVR(kernel='rbf',gamma=0.01,C=100)
svr.fit(X_train1,y_train1)
```

```python
y_pred1=svr.predict(X_test1)
from sklearn.metrics import r2_score
print(r2_score(y_test1,y_pred1))
```

```python
grid=GridSearchCV(svr,param_grid=params_grid,refit=True,cv=5,verbose=3)
grid.fit(X_train1,y_train1)
```
